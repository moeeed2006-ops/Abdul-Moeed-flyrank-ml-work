## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

Based on an initial run of the complete 30,000-page dataset, the core baseline numbers reveal a severe performance issue. Over 54% of the URLs match the criteria for traffic decay, and a massive percentage of these declining assets are stale, having gone untouched for over half a year. This proves that a predictive prioritization model is highly valuable over the next 7 weeks.

```python
import os
import sys
import subprocess
import pandas as pd
import numpy as np

# 1. Environment Setup (Ensures data is cloned into Colab)
IN_COLAB = "google.colab" in sys.modules
REPO_URL = "https://github.com/flyrank-bih/flyrank-ml-internship-starter"
REPO_DIR = "flyrank-ml-internship-starter"

if IN_COLAB:
    if not os.path.isdir(REPO_DIR):
        subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO_DIR], check=True)
    os.chdir(REPO_DIR)
else:
    while not os.path.isdir("data/raw") and os.getcwd() != "/":
        os.chdir("..")

# 2. Load the Starter Data
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# 3. Calculate the metrics
total_pages = len(df)
declining_pages = df["trend_direction"].str.lower().eq("down").sum()
pct_declining = (declining_pages / total_pages) * 100

stale_and_declining = df[(df["days_since_last_update"] >= 180) & (df["trend_direction"].str.lower() == "down")].shape[0]
pct_stale_declining = (stale_and_declining / declining_pages) * 100

print(f"Real Number 1 (Total Pages): {total_pages}")
print(f"Real Number 2 (Overall Declining Rate): {pct_declining:.1f}%")
print(f"Real Number 3 (Stale & Declining Pages): {pct_stale_declining:.1f}% of declining pages haven't been updated in 180+ days.")
```

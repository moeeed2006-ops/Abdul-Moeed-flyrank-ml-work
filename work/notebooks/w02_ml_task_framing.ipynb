## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

Answer:

A hardcoded rule (like days_since_last_update > 180) only captures a single dimension of content rot. Organic traffic performance patterns are non-linear and deeply relational. For instance, a highly visible page might start decaying after only 60 days if competitors target the same keyword, whereas a low-visibility baseline resource might remain stable for a year. An ML model tracks complex feature cross-interactions (such as the relationship between velocity of click loss, age, and broad query trends) that are too messy and volatile for an nested if-else statement to handle dynamically.
